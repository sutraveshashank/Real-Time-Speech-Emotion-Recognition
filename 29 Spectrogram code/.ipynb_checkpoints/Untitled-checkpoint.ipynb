```python
import numpy as np
from pydub import AudioSegment
from pydub.generators import Sine
import scipy.signal as signal
import matplotlib.pyplot as plt
import soundfile as sf
import sounddevice as sd
from IPython.display import Audio
```

```python


# Audio recording parameters
samplerate = 44100  # Sample rate in Hz
channels = 2        # Stereo audio
duration = 5        # Duration in seconds

def record_audio(filename):
    print("Recording started... Speak now!")
    
    # Record audio
    audio_data = sd.rec(int(samplerate * duration), samplerate=samplerate, channels=channels, dtype='float32')
    sd.wait()  # Wait for the recording to complete
    
    print("Recording finished! Playing the audio...")
        # Save as WAV file
    sf.write(filename, audio_data, samplerate)

    # Return the recorded audio for playback
    return audio_data

```

```python
# Record audio
filename1 = "audio_data1.wav" 
audio_data1 = record_audio(filename1)

# Play the recorded audio in Jupyter Notebook
Audio(audio_data1.T, rate=samplerate)
```

```python
# Record audio
filename2 = "audio_data2.wav" 
audio_data2 = record_audio(filename2)

# Play the recorded audio in Jupyter Notebook
Audio(audio_data2.T, rate=samplerate)
```

```python
# Plot the waveforms of the original, mixed, and separated audio
def plot_waveform(audio, title="Audio Waveform"):
    # Save the audio to a temporary file to read it using soundfile
    temp_filename = "temp_audio.wav"
    audio.export(temp_filename, format="wav")
    
    # Load the audio file
    audio_data, samplerate = sf.read(temp_filename)

    # Create time axis
    duration = len(audio_data) / samplerate
    time = np.linspace(0., duration, len(audio_data))

    plt.figure(figsize=(12, 4))
    
    # Check if the audio is stereo or mono
    if len(audio_data.shape) == 1:
        plt.plot(time, audio_data, label="Mono Audio", color="b")
    else:
        plt.plot(time, audio_data[:, 0], label="Left Channel", color="b", alpha=0.7)
        plt.plot(time, audio_data[:, 1], label="Right Channel", color="r", alpha=0.7)

    plt.title(title)
    plt.xlabel("Time (seconds)")
    plt.ylabel("Amplitude")
    plt.legend()
    plt.grid()
    plt.show()
```

Existing

```python
# Step 1: Load the audio files
spk1_audio = AudioSegment.from_wav("audio_data1.wav")
spk2_audio = AudioSegment.from_wav("audio_data2.wav")

# Ensure both audio files have the same sample rate, otherwise resample
if spk1_audio.frame_rate != spk2_audio.frame_rate:
    spk2_audio = spk2_audio.set_frame_rate(spk1_audio.frame_rate)

# Step 2: Concatenate the audio files one after another (not using '+')
spk1_samples = np.array(spk1_audio.get_array_of_samples())
spk2_samples = np.array(spk2_audio.get_array_of_samples())

# Concatenate the samples (one after the other)
mixed_samples = np.concatenate((spk1_samples, spk2_samples))

# Convert the concatenated samples back to an AudioSegment
mixed_audio = AudioSegment(
    mixed_samples.astype(np.int16).tobytes(),
    frame_rate=spk1_audio.frame_rate,
    sample_width=spk1_audio.sample_width,
    channels=spk1_audio.channels
)

# Step 3: Hybrid Adaptive Audio Equalizer (Bandpass Filtering)
def bandpass_filter(audio, lowcut, highcut, fs, order=5):
    nyquist = 0.5 * fs
    low = lowcut / nyquist
    high = highcut / nyquist
    b, a = signal.butter(order, [low, high], btype='band')
    samples = np.array(audio.get_array_of_samples())
    filtered_samples = signal.filtfilt(b, a, samples)
    filtered_audio = AudioSegment(
        filtered_samples.astype(np.int16).tobytes(),
        frame_rate=fs,
        sample_width=audio.sample_width,
        channels=audio.channels
    )
    return filtered_audio

# Define sample rate (assuming both files have the same sample rate now)
sample_rate = spk1_audio.frame_rate

# Apply bandpass filters for both speakers
spk1_separated = bandpass_filter(mixed_audio, 900, 1100, sample_rate)  # Speaker 1
spk2_separated = bandpass_filter(mixed_audio, 1400, 1600, sample_rate) # Speaker 2

# Step 4: Save the separated audio files
mixed_audio.export("Existing_Mixed.wav", format="wav")
spk1_separated.export("Existing_spk1_separated.wav", format="wav")
spk2_separated.export("Existing_spk2_separated.wav", format="wav")


# Plot waveforms
```

```python
plot_waveform(spk1_audio, "Speaker 1 Audio")
plot_waveform(spk2_audio, "Speaker 2 Audio")
plot_waveform(mixed_audio, "Mixed Audio")
```

```python
plot_waveform(spk1_audio, "Original Speaker 1 Audio")
plot_waveform(spk1_separated, "Separated Audio - Speaker 1")
```

```python
plot_waveform(spk2_audio, "Original Speaker 2 Audio")
plot_waveform(spk2_separated, "Separated Audio - Speaker 2")
```

Proposed

```python
# Step 1: Load the audio files
spk1_audio = AudioSegment.from_wav("audio_data1.wav")
spk2_audio = AudioSegment.from_wav("audio_data2.wav")

# Step 2: Ensure both signals have the same sample rate
if spk1_audio.frame_rate != spk2_audio.frame_rate:
    spk2_audio = spk2_audio.set_frame_rate(spk1_audio.frame_rate)

# Convert audio files to numpy arrays
spk1_samples = np.array(spk1_audio.get_array_of_samples())
spk2_samples = np.array(spk2_audio.get_array_of_samples())

# Step 3: Extend signal 1 to match the time scale of signal 2
if len(spk1_samples) < len(spk2_samples):
    spk1_samples = np.pad(spk1_samples, (0, len(spk2_samples) - len(spk1_samples)), 'constant')

# Step 4: Concatenate the signals (signal 1 first, followed by signal 2)
mixed_samples = np.concatenate((spk1_samples, spk2_samples))

# Step 5: Create a mixed audio object from the concatenated samples
mixed_audio = AudioSegment(
    mixed_samples.astype(np.int16).tobytes(),
    frame_rate=spk1_audio.frame_rate,
    sample_width=spk1_audio.sample_width,
    channels=spk1_audio.channels
)

# Step 6: Separate the mixed audio back into individual signals based on time scales
signal1_separated = mixed_samples[:len(spk1_samples)]
signal2_separated = mixed_samples[len(spk1_samples):]

# Step 7: Convert the separated signals back into AudioSegments
signal1_audio = AudioSegment(
    signal1_separated.astype(np.int16).tobytes(),
    frame_rate=spk1_audio.frame_rate,
    sample_width=spk1_audio.sample_width,
    channels=spk1_audio.channels
)

signal2_audio = AudioSegment(
    signal2_separated.astype(np.int16).tobytes(),
    frame_rate=spk1_audio.frame_rate,
    sample_width=spk1_audio.sample_width,
    channels=spk1_audio.channels
)

# Step 8: Save the separated audio files
mixed_audio.export("Proposed_Mixed.wav", format="wav")
signal1_audio.export("Proposed_spk1_separated.wav", format="wav")
signal2_audio.export("Proposed_spk2_separated.wav", format="wav")


# Plot waveforms of the original, mixed, and separated audio
```

```python
plot_waveform(spk1_audio, "Speaker 1 Audio")
plot_waveform(spk2_audio, "Speaker 2 Audio")
plot_waveform(mixed_audio, "Mixed Audio")
```

```python
plot_waveform(spk1_audio, "Original Speaker 1 Audio")
plot_waveform(signal1_audio, "Separated Audio - Speaker 1")
```

```python
plot_waveform(spk2_audio, "Original Speaker 2 Audio")
plot_waveform(signal2_audio, "Separated Audio - Speaker 2")
```
